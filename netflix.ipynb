```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import warnings as ws 
ws.filterwarnings("ignore")
```

```python
df = pd.read_csv(r"C:\Users\BCS\Downloads\netflix_titles.csv")
df.tail(20)
```

```python
release = df['release_year'].value_counts().idxmax()
df['release_year'] = df['release_year'].fillna(release)
```

```python
df['rating'] = df['rating'].fillna('R')
```

```python
#List all the Indian movies released in the year 2019.
df[df['release_year']== 2019.0].head(5)
```

```python
#List all the movies in such a manner that it shows shortest movies first
sorted_value = df.sort_values(by=df['duration'] ,ascending=True)
print(sorted_value)
```

```python
#Create a new data frame by adding a new attribute i.e. “HowOld”. The
#value of the newly added attribute can be calculated as “current year – release year
current_year = 2026
df['Howold'] = current_year - df['release_year']
df.head(5)
```

```python
#Write the newly created data frame into a csv file and name it as  “newNetflixData.csv”.
df.to_csv('newNetflixData.csv', index=False)
```
