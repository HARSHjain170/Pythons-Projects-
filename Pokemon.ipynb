```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt 
import warnings as ws 
ws.filterwarnings("ignore")
```

```python
df = pd.read_csv(r"C:\Users\BCS\Downloads\Pokemon.csv")
df.head(20)
```

```python
df['Type 1'] = df['Type 1'].fillna('Water')
```

```python
least_use = df['Type 2'].value_counts().idxmin()
df['Type 2'] = df['Type 2'].replace('Blank',least_use)
```

```python
avg_speed = df['Speed'].mean()
print(avg_speed)
df [df['Speed'] > avg_speed]
```

```python
df.sort_values(by='Atk Sp. ',ascending=False).head(5)
```

```python
df['Total'] = df['HP'] + df['Attack'] + df['Defense Sp. '] + df['Atk Sp. '] + df['Def'] + df['Speed']
```

```python
avg_total = df['Total'].mean()
avg_total
df[df['Total'] >= avg_total]
```

```python
total_number = df['Type 1'].value_counts()
total_number.plot(kind='bar')
plt.xticks(rotation =90)
plt.show()
```

```python
df['Legendary'] = df['Legendary'].replace({'False':'0',
                                          'True':'1'})
df.head(5)
```

```python
flying_pokemon = df[(df['Type 2']=='Flying') & (df['Speed'] > 100) ]
print(flying_pokemon.head(5))
```

```python
pokemon_fire = df[(df['Generation'] == 6) | (df['Type 1'] == 'Fire')]
print(pokemon_fire.head(5))
```

```python
df['Generation'].value_counts()
```

```python
Poison = df[(df['Type 2']== 'Poison')&(df['Generation'].isin([4,5,6]))]
print(Poison.head(5))
```
