```python
#Calculate the proportion of deaths per month
monthly_df['proportion_of_death_per_month'] = monthly_df['deaths']/monthly_df['births']
monthly_df.head(5)
```

```python
monthly_df.dtypes
```

```python
#Change the data type of "date" column from string to datatime
monthly_df['date'] = pd.to_datetime(monthly_df['date']
monthly_df.dtypes
```

```python
# Label the date at which handwashing started to "start_handwashing"
start_handwashing = pd.to_datetime('1847-06-01')
# Split monthly into before and after handwashing_start
before_handwashing = monthly_df[monthly_df['date'] < start_handwashing].head(5)
after_handwashing = monthly_df[monthly_df['date']>= start_handwashing].head(5)
print(before_handwashing.head(5))
print(after_handwashing.head(5))
```

```python
fig,ax = plt.subplots(figsize=(12,6))
x = before_handwashing['date']
y = before_handwashing['proportion_of_death_per_month']
plt.plot(x,y,color ='green')
plt.xlabel("Dates")
plt.ylabel("Proportion of deaths ")
plt.show()
```

```python
fig,ax = plt.subplots(figsize = (10,4))
x= before_handwashing["date"]
y= before_handwashing["proportion_of_death_per_month"]
plt.plot(x, y, color= "orange")
plt.title("Before Handwashing", fontsize=16)
plt.xlabel("Date", fontsize=14)
plt.ylabel("Proportion of Deaths", fontsize=14)
```
