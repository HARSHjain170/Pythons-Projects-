```python
quality_counts = wine_df['quality'].value_counts() 
plt.figure(figsize=(12,8))
plt.bar(quality_counts.index,quality_counts,color='pink')
plt.title("Count Plot of Quatiy",fontsize=14)
plt.xlabel("Quality Counts")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(12,8))
sns.swarmplot(x='quality',y='alcohol',data= wine_df,palette='viridis')
plt.title('Alcohol Wise Quality')
plt.xlabel('Quality')
plt.ylabel('Alcohol')
plt.show()
```

```python
wine_df['quality'] = wine_df['quality'].astype(str)
plt.figure(figsize=(12,8))
sns.violinplot(x='quality',y='alcohol',data=wine_df,palette='viridis',alpha=0.8)
plt.title("viollinplot for Quality and alcohol")
plt.xlabel('Quality')
plt.ylabel('Alcohol')
plt.show()
```

```python
sns.boxplot(x='quality',y='alcohol',data=wine_df)
plt.show()
```
